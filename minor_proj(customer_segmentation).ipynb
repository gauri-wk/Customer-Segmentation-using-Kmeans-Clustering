# Data Preprocessing

# 1.Handling missing values

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

# Feature Engineering

```python
# Age

df["Age"] = 2026 - df["Year_Birth"]
```

```python
df.head()
```

```python
# Customer Tenure Days

df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)
```

```python
df.head()
```

```python
reference_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
# Spending

df["Total_Spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
```

```python
# Children

df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
# Education

df["Education"].value_counts()
```

```python
df["Education"] = df["Education"].replace({
    "2n Cycle" : "UnderGraduate", "Basic" : "UnderGraduate",
    "Graduation" : "Graduate",
    "PhD" : "PostGraduate", "Master" : "PostGraduate",   
})
```

```python
df["Education"].value_counts()
```

```python
# Marital Status

df["Marital_Status"].value_counts()
```

```python
df["Living_Status"] = df["Marital_Status"].replace({
    "Single" : "Alone", "Divorced" : "Alone", "Widow" : "Alone", "Absurd" : "Alone", "YOLO" : "Alone",
    "Married" : "Partner", "Together" : "Partner"
})
```

```python
df["Living_Status"].value_counts()
```

```python
df.columns
```

```python
cols = ["ID", "Year_Birth","Marital_Status", "Kidhome", "Teenhome", "Dt_Customer"]

spending_cols =  ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]

columns_to_drop = cols + spending_cols

df_cleaned = df.drop(columns=columns_to_drop)
```

# Feature Encoding

```python
from sklearn.preprocessing import OneHotEncoder 
```

```python
ohe = OneHotEncoder()

cat_cols = ["Education", "Living_Status"]
enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
enc_df.head()
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols), enc_df], axis=1)
```

```python
df_encoded.shape
```

# Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = df_encoded
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

# Visualize

```python
from sklearn.decomposition import PCA
```

```python
# 2D
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_ratio_
```

```python
fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2])

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")
ax.set_title("3D Projection")
```

# Analyze K
# 1. ELbow method

```python
from sklearn.cluster import KMeans

wcss = []
for k in range(1, 11):
    kmeans=KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("K")
plt.ylabel("WCSS")
```

# Silhouette score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    scores.append(score)

#plot
plt.plot(range(2, 11), scores, marker='o')
plt.xlabel("K")
plt.ylabel("Silhouette score")
```

```python
# Combined plot - k value near the intersection of the both is to be selected

k_range = range(2, 11)

fig, ax1 = plt.subplots(figsize=(10, 8))

ax1.plot(k_range, wcss[:len(k_range)], marker='o', color='blue')
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker='x', color='red', linestyle='--')
ax2.set_ylabel("SS")
```

# Clustering

```python
# Kmeans

kmeans = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)

# plot
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels_kmeans)
```

```python
# Agglomerative

from sklearn.cluster import AgglomerativeClustering

agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)

# plot
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels_agg)
```

# Characterization of clusters

```python
X["Clusters"] = labels_agg
```
